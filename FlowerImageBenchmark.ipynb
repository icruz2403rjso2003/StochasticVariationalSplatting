```python
from sklearn.datasets import load_sample_image

import matplotlib.pyplot as plt

from ImageToPatches import ImageToPatches
```

```python
flower = load_sample_image('flower.jpg')
```

```python
plt.figure(figsize = (15, 7.5))

plt.imshow(flower, aspect = 'auto')

plt.axis('off')

plt.show()
```

```python
ITP = ImageToPatches(image = flower, patch_size = 100)

ITP.image_to_patches()
```

```python
print('Pixels da Imagem:\n')

print(ITP.pixel)
```

```python
print('Cores da Imagem:\n')

print(ITP.color)
```

```python
plt.figure(figsize = (15, 7.5))

plt.imshow(ITP.color.reshape((ITP.H, ITP.W, 3)), aspect = 'auto')

plt.axis('off')

plt.show()
```
